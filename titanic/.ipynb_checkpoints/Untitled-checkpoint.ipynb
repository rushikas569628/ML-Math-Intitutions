```python
#inspect if NAN are replaced with median 28
df.groupby(["pclass", "sex"])["age"].median()
```

Group-wise median imputation was applied using passenger class and sex. Although a spike remains around age 28, this is expected because several groups (2,female), (3,female), and (3,male) have a median age of 28. Therefore, many missing values in these groups were imputed with 28, creating a concentration around that age while still preserving group-specific age patterns better than global median imputation.

Missing values in Age were imputed using group-wise median imputation based on passenger class (pclass) and sex (sex). A concentration around age 28 remains because multiple groups have a median age of 28, resulting in many missing values being imputed with that value.


```python
df["pclass"].unique()
```

### Groupwise /Conditional Imputation
Because different groups may have different underlying distributions.
1. Find group

2. Calculate that group's median age

3. Replace NaN only within that group

### Missingness Indicator
This creates:
- 1 = Age was missing.
- 0 = Age was available.
Why?

Sometimes the fact that information is missing is itself meaningful.

For example, missing income information in a customer dataset might be associated with a particular customer group or data collection process.

### 2. Bivariate Analysis

#### A. Categorical vs Categorical

```python
pd.crosstab(
    df["sex"],
    df["survived"]
)
```

```python
pd.crosstab(df["sex"],df["survived"],normalize="index")*100
```

- Why use normalize="index"?

Because we want to calculate the percentage of survivors and non-survivors within each gender, not across the entire dataset.

```python
ax=sns.countplot(data=df,x="sex",hue="survived")
pct=pd.crosstab(df["sex"],df["survived"],normalize="index")*100
labels=[f"{pct.loc["male",0]:.1f}%",
       f"{pct.loc["male",1]:.1f}%",
       f"{pct.loc["female",0]:.1f}%",
        f"{pct.loc["female",1]:.1f}%"]

for bar, label in zip(ax.patches, labels):
    ax.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        label,
        ha="center"
    )

plt.title("Countplot of Sex by Survived")
#plt.savefig("Countplot of Gender by Survived")
plt.show()
```

#### B. Numerical vs categorical

```python
df.groupby("survived")["age"].agg([
    "count",
    "min",
    "median",
    "mean",
    "max"
])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

ax = sns.boxplot(data=df, x="survived", y="age")

# Get statistics for each survival group
stats = df.groupby("survived")["age"].describe()

for i, grp in enumerate(stats.index):

    q1 = stats.loc[grp, "25%"]
    median = stats.loc[grp, "50%"]
    q3 = stats.loc[grp, "75%"]
    mean = stats.loc[grp, "mean"]
    min_val = stats.loc[grp, "min"]
    max_val = stats.loc[grp, "max"]

    text = (
        f"Min: {min_val:.1f}\n"
        f"Q1: {q1:.1f}\n"
        f"Median: {median:.1f}\n"
        f"Mean: {mean:.1f}\n"
        f"Q3: {q3:.1f}\n"
        f"Max: {max_val:.1f}"
    )

    ax.text(
        i + 0.15,      # move text slightly right of box
        median,        # place near median
        text,
        fontsize=8
    )

plt.title("Age Distribution by Survival")
#plt.savefig("Age Distribution by Survival")
plt.show()
```

### Outlier Detection and treatment

##### 1. Boxplot method

```python
sns.boxplot(x=df["fare"])
plt.title("Fare outlier Detection")
#plt.savefig("Boxplot for Fare outlier Detection")
plt.show()
```

#### 2. IQR method

- IQR=Q3-Q1
  where Q1=25th percentile,Q3=75th percentile
- Lower Boundary=Q1-1.5(IQR)
- Upper Boundary=Q3+1.5(IQR)

| Method         | When to Use                                                                  |
|----------------|------------------------------------------------------------------------------|
| Keep           | Values are legitimate                                                       |
| Remove         | Values are confirmed errors or invalid observations                         |
| Capping        | Extreme values need to be limited for a justified analytical purpose        |
| Transformation | Distribution is highly skewed and a transformation makes analysis easier   |

```python
Q1=df["fare"].quantile(0.25)
Q3=df["fare"].quantile(0.75)
IQR=Q3-Q1
Lower_bound=Q1-1.5*IQR
Upper_bound=Q3+1.5*IQR
print(f"Lower_bound:{Lower_bound}")
print(f"Upper_bound:{Upper_bound}")
outliers=df[(df["fare"]<Lower_bound)|(df["fare"]>Upper_bound)]
print(outliers)
```

```python
df["Fare_log"] = np.log1p(df["fare"])
df["Fare_log"]
```

- Why log1p?

It calculates:
\[
log(1+x)
\]

It also handles zero values safely.

# Feature Engineering

Feature engineering means creating new meaningful columns or modifying existing columns to make the data more informative.

#### 1. Feature Creation

- Family size
- IsAlone
- Extract Title from Name
- Group rare Titles
- Extract deck from cabin
- Ticket Group Size

##### A. Family Size

```python
df["familysize"]=(df["sibsp"]+df["parch"]+1)
df["familysize"]
```

```python
df.columns
```

##### B. IsAlone

- Interpretation:
    - 1 = Travelling alone.
    - 0 = Travelling with family.

```python
df["isalone"]=np.where(df["familysize"]==1,1,0)
df["isalone"]
```

```python
df.columns
```

##### C. Extract title from Name

```python
df["title"] = df["name"].str.extract(
    r",\s*([^.]*)\.")
df["title"].value_counts()
```

##### D. Group rare Titles

```python
title_counts=df["title"].value_counts()
rare_titles=title_counts[title_counts<10].index
df["title"]=df["title"].replace(rare_titles,"Rare")
df["title"]
df["title"].value_counts()
```

##### E. Extract deck from Cabin

```python
df["deck"] = df["cabin"].str[0]
```

```python
df["deck"].value_counts(dropna=False)
```

```python
df["deck"] = df["cabin"].str[0].fillna("Unknown")
df["deck"]
```

| Problem               | Questions to Ask                                      | Possible Approach                              |
|-----------------------|-------------------------------------------------------|-------------------------------------------------|
| Missing data          | How much is missing? Why is it missing?               | Mean, median, mode, group-based imputation      |
| Duplicates            | Are these genuinely repeated records?                 | Remove confirmed duplicates                     |
| Wrong data types      | Does the data type match its meaning?                 | Type conversion                                  |
| Outliers              | Are these errors or legitimate extreme values?        | Retain, remove, cap, transform                   |
| Skewness              | Is the distribution asymmetric?                       | Median, log transformation                       |
| Categorical data      | Is there an order between categories?                 | One-hot or ordinal encoding                      |
| Numerical data        | Are ranges and units reasonable?                      | Validation, transformation, binning              |
| Feature creation      | Can existing information be combined or extracted?   | Aggregation, extraction, interaction             |
| Date columns          | Are there time-based patterns?                        | Extract year, month, weekday                     |
| High-cardinality columns | Are there too many unique categories?              | Grouping, extraction, domain-specific treatment  |

```python
import numpy as np

# 2 users (features: movies ratings for 5 movies)
user1 = np.array([5, 3, 0, 1, 4])
user2 = np.array([4, 2, 1, 1, 5])

# Dot product
dot = np.dot(user1, user2)

# Magnitudes
norm1 = np.linalg.norm(user1)
norm2 = np.linalg.norm(user2)

# Cosine similarity
cos_sim = dot / (norm1 * norm2)

print("Cosine Similarity:", cos_sim)
```
